# 16. Ligand ADMET / Drug-likeness Properties

Calculate QED, SA-score, Lipinski Ro5, Veber, and PAINS alerts for one or more ligands using RDKit.

> **Requires:** `pip install mdatools[admet]`

```python
# ============================================================
# CONFIG — edit only this cell
# ============================================================

# SMILES for the ligand(s) to evaluate
SMILES_DICT = {
    "aspirin":   "CC(=O)Oc1ccccc1C(=O)O",
    "ibuprofen":  "CC(C)Cc1ccc(cc1)C(C)C(=O)O",
    "caffeine":   "Cn1cnc2c1c(=O)n(c(=O)n2C)C",
}

# Set check_pains=True to detect PAINS structural alerts
CHECK_PAINS = True
# ============================================================
```

```python
from mdatools.analysis.admet import ADMETCalculator

calc = ADMETCalculator(check_pains=CHECK_PAINS)

results = {name: calc.from_smiles(smi, name=name) for name, smi in SMILES_DICT.items()}

for name, r in results.items():
    print(f"{name}: MW={r.mw:.1f} LogP={r.logp:.2f} QED={r.qed:.3f} "
          f"SA={r.sa_score:.2f} Ro5={r.lipinski_pass} PAINS={r.pains_alert}")
```

## Radar Chart (single molecule)

```python
import matplotlib.pyplot as plt
from mdatools.plotting import plot_admet_radar

for name, result in results.items():
    fig = plot_admet_radar(result)
    plt.show()
```

## Multi-compound comparison

```python
from mdatools.plotting import plot_admet_comparison

fig = plot_admet_comparison(results)
plt.show()
```

## Batch processing from SMILES list

```python
from mdatools.plotting import plot_qed_distribution

smiles_list = list(SMILES_DICT.values())
batch_df = calc.batch(smiles_list)
print(batch_df[["name", "mw", "logp", "qed", "sa_score", "lipinski_pass", "pains_alert"]])

fig = plot_qed_distribution(batch_df)
plt.show()
```
